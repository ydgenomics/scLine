# 基于`git clone`复制scLine代码试运行

## 1. 安装scLine

**云平台个性分析选联网模式**

```shell
git clone https://github.com/ydgenomics/scLine.git
cd scLine
nextflow run main.nf --help
```

因为我的`nextflow`安装在conda的base环境，故先激活conda环境，再运行`nextflow run main.nf --help`

## 2. 跑测试数据尝试检查环境配置

跑测试数据试试，先查看qc的示例代码，复制粘贴运行试试，注意修改真实测试数据地址

```shell
nextflow run main.nf --mode qc \
        --rawPaths '/data/work/scLine/input/sample1/sample1_raw|/data/work/scLine/input/sample2/sample2_raw' \
        --filterPaths '/data/work/scLine/input/sample1/sample1_filter|/data/work/scLine/input/sample2/sample2_filter' \
        --biosampleValues 'group1|group2' \
        --sampleValues 'sample1|sample2' \
        --runsoupx 'yes' \
        --outdir './results/qc'
```

运行时会在工作路径生成一个work目录，记录运行过程，可进入查看运行log，查看进程或检查报错

## 3. 结果文件

默认保存路径为工作路径下新建results目录。

work目录作为运行空间，存在缓存资源，如不需要过程信息，运行结束后使用`nextflow clean -f`清除缓存

过程信息见`*html`文件
